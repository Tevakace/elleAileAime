```python
import numpy as np


def mean_squared_error_gd(y, tx, initial_w, max_iters, gamma):
    y = np.asarray(y, dtype=float).reshape(-1)
    tx = np.asarray(tx, dtype=float)
    w = np.asarray(initial_w, dtype=float).reshape(-1).copy()

    for _ in range(max_iters):
        error = tx @ w - y
        gradient = tx.T @ error / y.size
        w = w - gamma * gradient

    error = y - tx @ w
    loss = 0.5 * np.mean(error ** 2)

    return w, float(loss)
```

```python
def mean_squared_error_sgd(y, tx, initial_w, max_iters, gamma):
    y = np.asarray(y, dtype=float).reshape(-1)
    tx = np.asarray(tx, dtype=float)
    w = np.asarray(initial_w, dtype=float).reshape(-1).copy()

    for _ in range(max_iters):
        n = np.random.randint(y.size)

        error = tx[n] @ w - y[n]
        gradient = tx[n] * error

        w = w - gamma * gradient

    error = y - tx @ w
    loss = 0.5 * np.mean(error ** 2)

    return w, float(loss)
```

```python
def least_squares(y, tx):
    y = np.asarray(y, dtype=float).reshape(-1)
    tx = np.asarray(tx, dtype=float)

    a = tx.T @ tx
    b = tx.T @ y

    w = np.linalg.solve(a, b)

    error = y - tx @ w
    loss = 0.5 * np.mean(error ** 2)

    return w, float(loss)
```

```python
def ridge_regression(y, tx, lambda_):
    y = np.asarray(y, dtype=float).reshape(-1)
    tx = np.asarray(tx, dtype=float)

    n, d = tx.shape

    a = tx.T @ tx + 2 * n * lambda_ * np.eye(d)
    b = tx.T @ y

    w = np.linalg.solve(a, b)

    error = y - tx @ w
    loss = 0.5 * np.mean(error ** 2)

    return w, float(loss)
```

```python
def logistic_regression(y, tx, initial_w, max_iters, gamma):
    y = np.asarray(y, dtype=float).reshape(-1)
    tx = np.asarray(tx, dtype=float)
    w = np.asarray(initial_w, dtype=float).reshape(-1).copy()

    for _ in range(max_iters):
        scores = tx @ w

        probabilities = np.exp(-np.logaddexp(0.0, -scores))

        gradient = tx.T @ (probabilities - y) / y.size
        w = w - gamma * gradient

    scores = tx @ w
    loss = np.mean(np.logaddexp(0.0, scores) - y * scores)

    return w, float(loss)
```

```python
def reg_logistic_regression(y, tx, lambda_, initial_w, max_iters, gamma):
    y = np.asarray(y, dtype=float).reshape(-1)
    tx = np.asarray(tx, dtype=float)
    w = np.asarray(initial_w, dtype=float).reshape(-1).copy()

    for _ in range(max_iters):
        scores = tx @ w

        probabilities = np.exp(-np.logaddexp(0.0, -scores))

        gradient = tx.T @ (probabilities - y) / y.size
        gradient = gradient + 2 * lambda_ * w

        w = w - gamma * gradient

    scores = tx @ w
    loss = np.mean(np.logaddexp(0.0, scores) - y * scores)

    return w, float(loss)
```
